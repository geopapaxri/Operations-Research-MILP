Ερώτημα 1(a): Μαθηματική Μοντελοποίηση
1. Σύνολα (Sets)

*   $N$: Το σύνολο των νοσηλευτών (Ν1 έως Ν10). Χωρίζεται στα υποσύνολα $N_{A}$(Ειδικότητα Α) και $N_{B}$ (Ειδικότητα Β).
*  $D$: Το σύνολο των ημερών της εβδομάδας, $d \in \{1, 2, ..., 7\}$ (όπου 1=Δευτέρα, 7=Κυριακή).
*  $S$: Το σύνολο των βαρδιών, $s \in \{\text{Π}, \text{Α}, \text{Ν}\}$ (Πρωινή, Απογευματινή, Νυχτερινή).

2. Παράμετροι (Parameters)

*   $C_{n,s}$: Το κόστος του νοσηλευτή $n$ για τη βάρδια $s$. Το κόστος της Νυχτερινής είναι 1,5 φορές το κόστος της Π/Α.

*   $Req_{d,s}$: Ο ελάχιστος απαιτούμενος αριθμός νοσηλευτών την ημέρα $d$ στη βάρδια $s$.

*   $U$: Το σύνολο των δεσμευμένων απουσιών (ζεύγη νοσηλευτή, ημέρας και βάρδιας που ο νοσηλευτής δεν μπορεί να εργαστεί).  

3. Μεταβλητές Απόφασης (Decision Variables)

*   $x_{n,d,s} \in \{0, 1\}$: Παίρνει την τιμή 1 αν ο νοσηλευτής $n$ εργάζεται την ημέρα $d$ στη βάρδια $s$, και 0 διαφορετικά.

*  $W_{max}, W_{min} \in \mathbb{R}^+$: Βοηθητικές συνεχείς μεταβλητές που αναπαριστούν τον μέγιστο και ελάχιστο συνολικό αριθμό βαρδιών οποιουδήποτε νοσηλευτή μέσα στην εβδομάδα.

 4. Αντικειμενική Συνάρτηση (Objective Function)
Στόχος είναι η ελαχιστοποίηση του συνολικού εβδομαδιαίου κόστους στελέχωσης:
  $$\min \sum_{n \in N} \sum_{d \in D} \sum_{s \in S} C_{n,s} x_{n,d,s}$$

  5. Περιορισμοί (Constraints) - Μέρος 1ο
*   Δεσμευμένες Απουσίες: Οι νοσηλευτές δεν μπορούν να εργαστούν στις βάρδιες που έχουν δηλώσει απουσία (Πίνακας 1).
  $$x_{n,d,s} = 0, \quad \forall (n,d,s) \in U$$
*   Ελάχιστες Απαιτήσεις Κάλυψης: Ο αριθμός των νοσηλευτών που εργάζονται κάθε ημέρα και βάρδια πρέπει να καλύπτει τις ελάχιστες απαιτήσεις του ΤΕΠ (Πίνακας 2).
  $$\sum_{n \in N} x_{n,d,s} \ge Req_{d,s}, \quad \forall d \in D, s \in S$$
  

6. Περιορισμοί (Constraints) - Μέρος 2ο (Κανόνες Λειτουργίας)

 Κανόνας (α): Σε κάθε βάρδια και ημέρα πρέπει να υπηρετεί τουλάχιστον ένας νοσηλευτής Ειδικότητας Α.
  $$\sum_{n \in N_{A}} x_{n,d,s} \ge 1, \quad \forall d \in D, s \in S$$
 Κανόνας (β): Κάθε νοσηλευτής εργάζεται το πολύ μία βάρδια ανά ημέρα.
  $$\sum_{s \in S} x_{n,d,s} \le 1, \quad \forall n \in N, d \in D$$Επίσης, ο εβδομαδιαίος αριθμός βαρδιών του πρέπει να είναι από 3 έως 5.
  $$3 \le \sum_{d \in D} \sum_{s \in S} x_{n,d,s} \le 5, \quad \forall n \in N$$Κανόνας (γ): Μετά από Νυχτερινή βάρδια, απαιτούνται τουλάχιστον 16 ώρες ανάπαυσης. Μια νυχτερινή λήγει στις 07:00, άρα ο νοσηλευτής δεν μπορεί να δουλέψει την Πρωινή ή την Απογευματινή της επόμενης ημέρας.
  $$x_{n,d,N} + x_{n,d+1,\Pi} + x_{n,d+1,A} \le 1, \quad \forall n \in N, d \in \{1, 2, ..., 6\}$$Κανόνας (δ): Δεν επιτρέπονται τρεις ή περισσότερες διαδοχικές Νυχτερινές βάρδιες (άρα το άθροισμα σε οποιοδήποτε παράθυρο 3 ημερών είναι το πολύ 2).
  $$x_{n,d,N} + x_{n,d+1,N} + x_{n,d+2,N} \le 2, \quad \forall n \in N, d \in \{1, 2, ..., 5\}$$Κανόνας (ε): Η διαφορά βαρδιών (φόρτου) μεταξύ οποιωνδήποτε δύο νοσηλευτών δεν υπερβαίνει το 1. Χρησιμοποιούμε τις βοηθητικές μεταβλητές $W_{max}$ και $W_{min}$:
  $$\sum_{d \in D} \sum_{s \in S} x_{n,d,s} \le W_{max}, \quad \forall n \in N$$$$\sum_{d \in D} \sum_{s \in S} x_{n,d,s} \ge W_{min}, \quad \forall n \in N$$$$W_{max} - W_{min} \le 1$$Κανόνας (στ): Το πολύ μία Νυχτερινή βάρδια το Σαββατοκύριακο (ημέρες 6 και 7).
  $$x_{n,6,N} + x_{n,7,N} \le 1, \quad \forall n \in N$$

```python
import pulp
```

```python
# ==========================================
# 1. PROBLEM DATA (Based on AEM: 11177 -> a1=1, a2=7, a3=7)
# ==========================================

days = list(range(1, 8)) # 1:Monday ... 7:Sunday
shifts = ['Π', 'Α', 'Ν'] # Π: Morning, Α: Afternoon, Ν: Night
nurses = ['N1', 'N2', 'N3', 'N4', 'N5', 'N6', 'N7', 'N8', 'N9', 'N10']

# Specialties (Type A and Type B)
type_A = ['N1', 'N2', 'N5', 'N6', 'N9']
type_B = ['N3', 'N4', 'N7', 'N8', 'N10']

# Cost for Morning/Afternoon shift (calculated based on AEM)
cost_PA = {
    'N1': 90, 'N2': 85, 'N3': 70, 'N4': 67, 'N5': 72,
    'N6': 95, 'N7': 83, 'N8': 71, 'N9': 88, 'N10': 84
}

# Unavailabilities: (Nurse, Day, Shift)
unavailability = [
    ('N1', 3, 'Π'), ('N1', 3, 'Α'),
    ('N3', 5, 'Ν'),
    ('N5', 1, 'Π'),
    ('N6', 6, 'Π'), ('N6', 6, 'Α'), ('N6', 6, 'Ν'),
    ('N8', 7, 'Ν'),
    ('N9', 2, 'Ν'), ('N9', 3, 'Ν')
]

# Minimum requirements per Day and Shift (Req[d][s])
req = {
    1: {'Π': 3, 'Α': 2, 'Ν': 2},
    2: {'Π': 3, 'Α': 2, 'Ν': 2},
    3: {'Π': 3, 'Α': 2, 'Ν': 2},
    4: {'Π': 3, 'Α': 2, 'Ν': 2},
    5: {'Π': 3, 'Α': 2, 'Ν': 2},
    6: {'Π': 2, 'Α': 2, 'Ν': 1},
    7: {'Π': 2, 'Α': 2, 'Ν': 1}
}
```

```python
# ==========================================
# 2. MILP MODEL INITIALIZATION & VARIABLES
# ==========================================

# Create the LP problem (Minimization)
prob = pulp.LpProblem("Hospital_Shift_Scheduling", pulp.LpMinimize)

# --- Decision Variables ---
# x[n][d][s]: 1 if nurse n works on day d in shift s, 0 otherwise
x = pulp.LpVariable.dicts("x", (nurses, days, shifts), cat='Binary')

# Auxiliary continuous variables for Rule (e) - min/max workload
W_max = pulp.LpVariable("W_max", lowBound=0, cat='Continuous')
W_min = pulp.LpVariable("W_min", lowBound=0, cat='Continuous')
```

```python
# ==========================================
# 3. OBJECTIVE FUNCTION & CONSTRAINTS
# ==========================================

# --- Objective Function ---
# The cost for the Night shift ('Ν') is 1.5 times the Morning/Afternoon cost
prob += pulp.lpSum(
    cost_PA[n] * x[n][d][s] if s != 'Ν' else 1.5 * cost_PA[n] * x[n][d][s]
    for n in nurses for d in days for s in shifts
), "Total_Cost"

# --- Constraints ---

# 1. Unavailabilities (Cannot work on blocked shifts)
for n, d, s in unavailability:
    prob += x[n][d][s] == 0, f"Unavail_{n}_{d}_{s}"

# 2. Minimum Coverage Requirements
for d in days:
    for s in shifts:
        prob += pulp.lpSum(x[n][d][s] for n in nurses) >= req[d][s], f"Req_{d}_{s}"

# 3. Rule (a): At least 1 Type A nurse per shift/day
for d in days:
    for s in shifts:
        prob += pulp.lpSum(x[n][d][s] for n in type_A) >= 1, f"TypeA_{d}_{s}"

# 4. Rule (b) & (e): Shifts per day/week and linking with W_max/W_min
for n in nurses:
    # Maximum 1 shift per day
    for d in days:
        prob += pulp.lpSum(x[n][d][s] for s in shifts) <= 1, f"Max1_per_day_{n}_{d}"

    # Between 3 to 5 shifts per week
    total_shifts = pulp.lpSum(x[n][d][s] for d in days for s in shifts)
    prob += total_shifts >= 3, f"Min3_{n}"
    prob += total_shifts <= 5, f"Max5_{n}"

    # Link total shifts to W_max and W_min (Rule e)
    prob += total_shifts <= W_max, f"Wmax_link_{n}"
    prob += total_shifts >= W_min, f"Wmin_link_{n}"

# Difference between maximum and minimum workload (Rule e)
prob += W_max - W_min <= 1, "Max_Diff_1"

# 5. Rule (c): At least 16 hours of rest after a Night shift
for n in nurses:
    for d in range(1, 7): # Monday to Saturday
        prob += x[n][d]['Ν'] + x[n][d+1]['Π'] + x[n][d+1]['Α'] <= 1, f"Rest_{n}_{d}"

# 6. Rule (d): No 3 consecutive Night shifts
for n in nurses:
    for d in range(1, 6): # Monday to Friday
        prob += x[n][d]['Ν'] + x[n][d+1]['Ν'] + x[n][d+2]['Ν'] <= 2, f"No3Nights_{n}_{d}"

# 7. Rule (f): Maximum 1 Night shift during the weekend
for n in nurses:
    prob += x[n][6]['Ν'] + x[n][7]['Ν'] <= 1, f"WeekendNights_{n}"
```

```python
# ==========================================
# 4. SOLVE AND PRINT RESULTS
# ==========================================

# Solve the model
prob.solve()

# Print the results
print("Solver Status:", pulp.LpStatus[prob.status])
print(f"Total Weekly Cost: € {pulp.value(prob.objective):.2f}\n")

print("--- OPTIMAL SHIFT SCHEDULE ---")
day_names = {1:'Δευτ.', 2:'Τρί.', 3:'Τετ.', 4:'Πέμ.', 5:'Παρ.', 6:'Σάβ.', 7:'Κυρ.'}
for d in days:
    print(f"\n{day_names[d]}:")
    for s in shifts:
        working = [n for n in nurses if pulp.value(x[n][d][s]) == 1]
        print(f"  Βάρδια {s}: {', '.join(working)}")

print("\n--- NURSE WORKLOAD (Number of Shifts) ---")
for n in nurses:
    shifts_worked = sum(pulp.value(x[n][d][s]) for d in days for s in shifts)
    print(f"{n}: {int(shifts_worked)} shifts")
```

Ερώτημα 1(c): Ανάλυση Περιορισμού Κανόνα (δ)
1. Είναι σωστός ο περιορισμός του συμφοιτητή;

Όχι, ο περιορισμός $x_{n,d,N} + x_{n,d+1,N} \le 1$ δεν είναι σωστός. Το σφάλμα έγκειται στο ότι αυτή η εξίσωση απαγορεύει τις δύο διαδοχικές Νυχτερινές βάρδιες, ενώ ο κανόνας (δ) αναφέρει ρητά ότι απαγορεύονται οι τρεις ή περισσότερες (δηλαδή οι δύο διαδοχικές επιτρέπονται).  Ως αντιπαραδειγμα, έστω ότι ο νοσηλευτής Ν1 εργάζεται Νυχτερινή βάρδια τη Δευτέρα ($d=1$) και την Τρίτη ($d=2$), και ξεκουράζεται την Τετάρτη ($d=3$). Αυτό είναι ένα καθόλα νόμιμο πρόγραμμα σύμφωνα με τον κανόνα (δ). Σε αυτή την περίπτωση έχουμε:

*   $x_{N1, 1, N} = 1$

*   $x_{N1, 2, N} = 1$

Αν εφαρμόσουμε τον περιορισμό του συμφοιτητή για $d=1$:
$x_{N1, 1, N} + x_{N1, 2, N} = 1 + 1 = 2$.
Όμως το 2 δεν είναι $\le 1$, συνεπώς το μοντέλο του συμφοιτητή θα απέρριπτε λανθασμένα αυτό το έγκυρο πρόγραμμα. Για να επιτρέψουμε το πολύ δύο διαδοχικές Νυχτερινές, πρέπει σε οποιοδήποτε "παράθυρο" 3 συνεχόμενων ημερών, ο αριθμός των Νυχτερινών βαρδιών να μην ξεπερνά το 2.
Ο ορθός περιορισμός είναι:
  $$x_{n,d,N} + x_{n,d+1,N} + x_{n,d+2,N} \le 2, \quad \forall n \in N, d \in \{1, 2, 3, 4, 5\}


Αν η εβδομάδα θεωρηθεί κυκλική (η Κυριακή ακολουθείται από τη Δευτέρα), τότε πρέπει να ελέγξουμε και τα "παράθυρα" 3 ημερών που περνούν από το τέλος της μίας εβδομάδας στην αρχή της επόμενης. Συγκεκριμένα, πρέπει να προσθέσουμε δύο επιπλέον περιορισμούς για κάθε νοσηλευτή:  Για το παράθυρο Σάββατο - Κυριακή - Δευτέρα:$$x_{n,6,N} + x_{n,7,N} + x_{n,1,N} \le 2, \quad \forall n \in N$$
Για το παράθυρο Κυριακή - Δευτέρα - Τρίτη:$$x_{n,7,N} + x_{n,1,N} + x_{n,2,N} \le 2, \quad \forall n \in N$$

Ερώτημα 2(a): Μαθηματική Μοντελοποίηση (MILP)
1.

*  Σύνολα (Sets)$V$: Το σύνολο όλων των κόμβων του δικτύου, $V = \{S, A, B, C, D, H, U, P, G\}$.
*   $K$: Το σύνολο των κρίσιμων κόμβων προορισμού, $K = \{H, U, P, G\}$.

*   $T$: Το σύνολο των χρονικών περιόδων, $T = \{1, 2, 3, 4\}$.
*   $E$: Το σύνολο των φυσικών ζεύξεων. Κάθε ζεύξη αναπαρίσταται ως δύο κατευθυνόμενα τόξα $(i,j)$ και $(j,i)$.

*   $E_{dam} \subset E$: Το υποσύνολο των αρχικά κατεστραμμένων φυσικών ζεύξεων ($e_4, e_7, e_{10}, e_{12}$).
*  $M$: Το σύνολο των διαθέσιμων προσωρινών ζεύξεων ($m_1, m_2, m_3$).

* $C$: Το σύνολο των συνεργείων επισκευής, $C = \{1, 2\}$.

2. Παράμετροι (Parameters)
*   $Dem_{k,t}$: Η ζήτηση του κρίσιμου κόμβου $k$ την περίοδο $t$.

*   $Pen_k$: Η ποινή ανά μονάδα μη εξυπηρετούμενης ζήτησης για τον κόμβο $k$.
*   $u_e$: Η χωρητικότητα της φυσικής ζεύξης $e \in E$.$u_m$: Η χωρητικότητα της προσωρινής ζεύξης $m \in M$.

*   $\tau_e$: Η διάρκεια επισκευής (σε περιόδους) της κατεστραμμένης ζεύξης $e \in E_{dam}$.
*   $c^r_e$: Το εφάπαξ κόστος επισκευής της ζεύξης $e \in E_{dam}$.
*   $c^m_m$: Το κόστος ανά περίοδο λειτουργίας της προσωρινής ζεύξης $m \in M$.

3. Μεταβλητές Απόφασης (Decision Variables)
*   $f_{k,i,j,t} \ge 0$: Η ροή με προορισμό τον κόμβο $k$, στο κατευθυνόμενο τόξο $(i,j)$ κατά την περίοδο $t$.

*   $unsat_{k,t} \ge 0$: Η μη εξυπηρετούμενη ζήτηση του κρίσιμου κόμβου $k$ κατά την περίοδο $t$.
*   $w_{c,e,t} \in \{0, 1\}$: $1$ αν το συνεργείο $c$ ξεκινήσει την επισκευή της ζεύξης $e$ την περίοδο $t$, $0$ διαφορετικά.$y_{e,t} \in \{0, 1\}$: $1$ αν η φυσική ζεύξη $e$ είναι διαθέσιμη προς χρήση κατά την περίοδο $t$, $0$ διαφορετικά.
*   $z_{m,t} \in \{0, 1\}$: $1$ αν η προσωρινή ζεύξη $m$ είναι ενεργή κατά την περίοδο $t$, $0$ διαφορετικά.

4. Αντικειμενική Συνάρτηση-Ελαχιστοποίηση του συνολικού κόστους (ποινές + επισκευές + προσωρινές ζεύξεις + δρομολόγηση):
$$\min \sum_{t \in T} \sum_{k \in K} Pen_{k} \cdot unsat_{k,t} + \sum_{c \in C} \sum_{e \in E_{dam}} \sum_{t \in T} c_{e}^{r} \cdot w_{c,e,t} + \sum_{m \in M} \sum_{t \in T} c_{m}^{m} \cdot z_{m,t} + \sum_{t \in T} \sum_{k \in K} \sum_{(i,j) \in E \cup M} f_{k,i,j,t}$$
5. Περιορισμοί

α)   Διατήρηση Ροής: Αν $i = S$ (Πηγή):$$\sum_{j} f_{k,S,j,t} - \sum_{j} f_{k,j,S,t} = Dem_{k,t} - unsat_{k,t}$$Αν $i = k$ (Προορισμός):$$\sum_{j} f_{k,k,j,t} - \sum_{j} f_{k,j,k,t} = -(Dem_{k,t} - unsat_{k,t})$$Αν $i \neq S, k$ (Ενδιάμεσοι κόμβοι):$$\sum_{j} f_{k,i,j,t} - \sum_{j} f_{k,j,i,t} = 0$$Επίσης: $0 \le unsat_{k,t} \le Dem_{k,t}$

β)Κοινή Χωρητικότητα Ζεύξεων:

a)Φυσικές: $\sum_{k \in K} (f_{k,i,j,t} + f_{k,j,i,t}) \le u_{e} \cdot y_{e,t}, \quad \forall e \in E, t \in T$

b)Προσωρινές: $\sum_{k \in K} (f_{k,i,j,t} + f_{k,j,i,t}) \le u_{m} \cdot z_{m,t}, \quad \forall m \in M, t \in T$

γ)Όριο Προσωρινών Ζεύξεων:$$\sum_{m \in M} z_{m,t} \le 2, \quad \forall t \in T$$

δ)Διαθεσιμότητα Φυσικών Ζεύξεων: Αν $e \notin E_{dam}$: $y_{e,t} = 1, \quad \forall t \in T$Αν $e \in E_{dam}$: $y_{e,t} \le \sum_{c \in C} \sum_{t'=1}^{t-\tau_{e}} w_{c,e,t'}, \quad \forall e \in E_{dam}, t \in T$

ε)Χρονισμός και Όρια Επισκευών:Κάθε βλάβη επισκευάζεται το πολύ 1 φορά:$$\sum_{c \in C} \sum_{t \in T} w_{c,e,t} \le 1, \quad \forall e \in E_{dam}$$Κάθε συνεργείο σε 1 βλάβη τη φορά:$$\sum_{e \in E_{dam}} \sum_{t' = \max(1, t - \tau_{e} + 1)}^{t} w_{c,e,t'} \le 1, \quad \forall c \in C, t \in T$$

```python
import pulp

# ==========================================
# 1. PROBLEM 2 DATA (Based on AEM: 11177)
# ==========================================

periods = [1, 2, 3, 4]
nodes = ['S', 'A', 'B', 'C', 'D', 'H', 'U', 'P', 'G']
critical = ['H', 'U', 'P', 'G']
crews = [1, 2]

# Capacities (u_e) of Physical Links
u_e = {
    'e1': 62, 'e2': 69, 'e3': 28, 'e4': 49, 'e5': 44, 'e6': 39, 'e7': 41,
    'e8': 38, 'e9': 37, 'e10': 33, 'e11': 27, 'e12': 35, 'e13': 26, 'e14': 37
}

# Endpoints of Physical Links
phys_links = {
    'e1': ('S','A'), 'e2': ('S','B'), 'e3': ('A','B'), 'e4': ('A','C'),
    'e5': ('B','C'), 'e6': ('B','D'), 'e7': ('C','H'), 'e8': ('C','U'),
    'e9': ('D','U'), 'e10': ('D','P'), 'e11': ('H','P'), 'e12': ('U','G'),
    'e13': ('B','G'), 'e14': ('C','P')
}

# Damaged Links Information (tau: duration of repair, cr: cost of repair)
damaged_info = {
    'e4': {'tau': 2, 'cr': 730},
    'e7': {'tau': 1, 'cr': 830},
    'e10': {'tau': 1, 'cr': 602},
    'e12': {'tau': 2, 'cr': 590}
}
damaged_edges = list(damaged_info.keys())
normal_edges = [e for e in phys_links.keys() if e not in damaged_edges]

# Temporary Links (Endpoints, Capacities, and Cost per period)
temp_links = {'m1': ('S','H'), 'm2': ('S','P'), 'm3': ('A','G')}
u_m = {'m1': 17, 'm2': 22, 'm3': 25}
c_m = {'m1': 955, 'm2': 985, 'm3': 900}

# Demand per critical node and period
demand = {
    'H': {1: 23, 2: 35, 3: 21, 4: 21},
    'U': {1: 28, 2: 31, 3: 30, 4: 34},
    'P': {1: 24, 2: 22, 3: 27, 4: 30},
    'G': {1: 13, 2: 19, 3: 28, 4: 26}
}

# Penalty per unit of unsatisfied demand
penalty = {'H': 165, 'U': 123, 'P': 165, 'G': 78}
```

```python
# ==========================================
# 2. MILP MODEL AND DECISION VARIABLES
# ==========================================

# Create the LP problem (Minimization)
prob2 = pulp.LpProblem("Network_Restoration", pulp.LpMinimize)

# Create directed arcs for both directions of each link
arcs_e = {e: [(i, j), (j, i)] for e, (i, j) in phys_links.items()}
arcs_m = {m: [(i, j), (j, i)] for m, (i, j) in temp_links.items()}
all_arcs = []
for e in arcs_e.values(): all_arcs.extend(e)
for m in arcs_m.values(): all_arcs.extend(m)

# -- Continuous Variables --
# Flow f[k][arc][t]: Flow for critical node k on directed arc during period t
f = pulp.LpVariable.dicts("f", (critical, all_arcs, periods), lowBound=0, cat='Continuous')
# Unsatisfied demand unsat[k][t]
unsat = pulp.LpVariable.dicts("unsat", (critical, periods), lowBound=0, cat='Continuous')

# -- Binary Variables --
# y[e][t]: 1 if physical link e is available during period t
y = pulp.LpVariable.dicts("y", (phys_links.keys(), periods), cat='Binary')
# z[m][t]: 1 if temporary link m is active during period t
z = pulp.LpVariable.dicts("z", (temp_links.keys(), periods), cat='Binary')
# w[c][e][t]: 1 if crew c starts repairing link e at the beginning of period t
w = pulp.LpVariable.dicts("w", (crews, damaged_edges, periods), cat='Binary')

# Constraint: Do not allow repair starts that would finish after the horizon T=4
for c in crews:
    for e in damaged_edges:
        tau = damaged_info[e]['tau']
        for t in periods:
            # If the repair takes 'tau' periods, it must start by 4 - tau + 1 at the latest
            if t > 4 - tau + 1:
                w[c][e][t].bounds(0, 0)
```

```python
# ==========================================
# 3. OBJECTIVE FUNCTION AND CONSTRAINTS
# ==========================================

# -- Objective Function --
cost_pen = pulp.lpSum(penalty[k] * unsat[k][t] for k in critical for t in periods)
cost_rep = pulp.lpSum(damaged_info[e]['cr'] * w[c][e][t] for c in crews for e in damaged_edges for t in periods)
cost_temp = pulp.lpSum(c_m[m] * z[m][t] for m in temp_links for t in periods)
cost_route = pulp.lpSum(f[k][arc][t] for k in critical for arc in all_arcs for t in periods)

prob2 += cost_pen + cost_rep + cost_temp + cost_route, "Total_Cost"

# -- Constraints --
for t in periods:
    # 1. Maximum of 2 temporary links active per period
    prob2 += pulp.lpSum(z[m][t] for m in temp_links) <= 2, f"Max_Temp_{t}"

    for k in critical:
        # Unsatisfied demand cannot exceed total demand
        prob2 += unsat[k][t] <= demand[k][t], f"Unsat_Max_{k}_{t}"

        # 2. Flow Conservation Constraints
        for node in nodes:
            outflow = pulp.lpSum(f[k][(u,v)][t] for (u,v) in all_arcs if u == node)
            inflow = pulp.lpSum(f[k][(u,v)][t] for (u,v) in all_arcs if v == node)

            if node == 'S': # Source node
                prob2 += outflow - inflow == demand[k][t] - unsat[k][t], f"Flow_S_{k}_{node}_{t}"
            elif node == k: # Destination (critical) node
                prob2 += outflow - inflow == -(demand[k][t] - unsat[k][t]), f"Flow_Dest_{k}_{node}_{t}"
            else:           # Intermediate nodes
                prob2 += outflow - inflow == 0, f"Flow_Inter_{k}_{node}_{t}"

    # 3. Physical Links Capacity Constraint
    for e, arcs in arcs_e.items():
        prob2 += pulp.lpSum(f[k][arcs[0]][t] + f[k][arcs[1]][t] for k in critical) <= u_e[e] * y[e][t], f"Cap_Phys_{e}_{t}"

    # 4. Temporary Links Capacity Constraint
    for m, arcs in arcs_m.items():
        prob2 += pulp.lpSum(f[k][arcs[0]][t] + f[k][arcs[1]][t] for k in critical) <= u_m[m] * z[m][t], f"Cap_Temp_{m}_{t}"

    # 5. Availability of normal (undamaged) physical links (always 1)
    for e in normal_edges:
        prob2 += y[e][t] == 1, f"Avail_Norm_{e}_{t}"

    # 6. Availability of damaged links (requires completed repair)
    for e in damaged_edges:
        tau = damaged_info[e]['tau']
        # Link is available only if repair started at a time t_prime <= t - tau
        prob2 += y[e][t] <= pulp.lpSum(w[c][e][t_prime] for c in crews for t_prime in periods if t_prime <= t - tau), f"Avail_Dam_{e}_{t}"

# 7. Each damaged link is repaired at most once
for e in damaged_edges:
    prob2 += pulp.lpSum(w[c][e][t] for c in crews for t in periods) <= 1, f"Max_1_Rep_{e}"

# 8. Each crew works on at most one repair at any given time (no overlapping)
for c in crews:
    for t in periods:
        active_repairs = pulp.lpSum(
            w[c][e][t_prime]
            for e in damaged_edges
            for t_prime in periods
            if t_prime <= t and t_prime + damaged_info[e]['tau'] - 1 >= t
        )
        prob2 += active_repairs <= 1, f"Crew_Cap_{c}_{t}"
```

```python
# ==========================================
# 4. SOLVE MODEL AND PRINT RESULTS
# ==========================================

# Solve the MILP problem
prob2.solve()

# Output the status and optimal objective value
print("Κατάσταση Επίλυσης:", pulp.LpStatus[prob2.status])
print(f"Συνολικό Κόστος: € {pulp.value(prob2.objective):.2f}\n")

# Print the repair schedule
print("--- ΠΡΟΓΡΑΜΜΑ ΕΠΙΣΚΕΥΩΝ ---")
for t in periods:
    for e in damaged_edges:
        for c in crews:
            if pulp.value(w[c][e][t]) == 1:
                print(f"Περίοδος {t}: Το Συνεργείο {c} ξεκινά επισκευή της ζεύξης {e} (Διάρκεια: {damaged_info[e]['tau']} περ.)")

# Print the active temporary links per period
print("\n--- ΕΝΕΡΓΕΣ ΠΡΟΣΩΡΙΝΕΣ ΖΕΥΞΕΙΣ ---")
for t in periods:
    active = [m for m in temp_links if pulp.value(z[m][t]) == 1]
    print(f"Περίοδος {t}: {', '.join(active) if active else 'Καμία'}")

# Print demand satisfaction per critical node
print("\n--- ΖΗΤΗΣΗ (Εξυπηρετούμενη / Συνολική) ---")
for t in periods:
    print(f"Περίοδος {t}:")
    for k in critical:
        u_val = pulp.value(unsat[k][t])
        s_val = demand[k][t] - u_val
        print(f"  Κόμβος {k}: {s_val:.1f}/{demand[k][t]} (Μη εξυπηρετούμενη: {u_val:.1f})")

# Print cost breakdown
print("\n--- ΑΝΑΛΥΣΗ ΚΟΣΤΟΥΣ ---")
print(f"Ποινές Ζήτησης: € {pulp.value(cost_pen):.2f}")
print(f"Επισκευές: € {pulp.value(cost_rep):.2f}")
print(f"Προσωρινές Ζεύξεις: € {pulp.value(cost_temp):.2f}")
print(f"Δρομολόγηση Ροής: € {pulp.value(cost_route):.2f}")
```

Ανάλυση Σφάλματος Μοντελοποίησης Χωρητικότητας

Ο περιορισμός του συμφοιτητή:
  $$f_{k,i,j,t} + f_{k,j,i,t} \le u_{e} z_{e,t}$$είναι λάθος διότι εφαρμόζεται χωριστά (independently) για τον κάθε κρίσιμο κόμβο προορισμού $k$. Η χωρητικότητα $u_{e}$ μιας φυσικής ζεύξης είναι ένας κοινός πόρος που πρέπει να μοιράζεται ταυτόχρονα από όλες τις ροές που τη διασχίζουν, ανεξαρτήτως του προορισμού τους. Με τη διατύπωση του συμφοιτητή, το μοντέλο επιτρέπει σε κάθε κόμβο να καταναλώσει από μόνος του ολόκληρη τη χωρητικότητα της ζεύξης. Στην πράξη, αυτό πολλαπλασιάζει λανθασμένα τη διαθέσιμη χωρητικότητα επί τον αριθμό των κρίσιμων κόμβων.  

  Ως αντιπαράδειγμα,
έστω μια φυσική ζεύξη $e$ (π.χ. μεταξύ των κόμβων A και B), η οποία είναι ενεργή/διαθέσιμη (άρα $z_{e,t} = 1$) και έχει μέγιστη χωρητικότητα 10 μονάδες ($u_{e} = 10$). Έστω επίσης ότι έχουμε μόνο δύο κρίσιμους κόμβους προορισμού στο δίκτυο, τους $k_{1}$ και $k_{2}$.  Αν το μοντέλο αποφασίσει να στείλει ταυτόχρονα 10 μονάδες ροής με προορισμό τον $k_{1}$ και 10 μονάδες ροής με προορισμό τον $k_{2}$ μέσα από την ίδια ακριβώς ζεύξη:

α)Ελέγχοντας τον περιορισμό του συμφοιτητή για τον $k_{1}$ έχουμε: 10 $\le$ 10 (Ισχύει)

β)Ελέγχοντας τον περιορισμό του συμφοιτητή για τον $k_{2}$ έχουμε: 10 $\le$ 10 (Ισχύει)

Ο solver του συμφοιτητή θα δεχόταν αυτή τη δρομολόγηση ως απόλυτα έγκυρη. Όμως, η πραγματική συνολική ροή που διασχίζει τη ζεύξη A-B είναι 20 (10 + 10). Το νούμερο αυτό υπερβαίνει τη φυσική χωρητικότητα των 10 μονάδων, πράγμα που σημαίνει ότι ο περιορισμός απέτυχε να αποτρέψει την "υπερφόρτωση" του δικτύο.


Για να είναι σωστό το μοντέλο, πρέπει να αθροίσουμε τις ροές όλων των κρίσιμων κόμβων $k \in K$ και έπειτα να φράξουμε το συνολικό άθροισμα με τη διαθέσιμη χωρητικότητα της ζεύξης:
  $$\sum_{k \in K} (f_{k,i,j,t} + f_{k,j,i,t}) \le u_{e} y_{e,t}$$
